# ⚖️ DLKR-LJP V5.0: Dynamic Legal Knowledge Retrieval for Legal Judgment Prediction
### Dual-Code Indian Criminal Justice System (IPC 1860 & BNS 2023)

**Core Technical Highlights:**
1. **Dynamic Retrieval Module (DRM)**: Inference-time dense & lexical retrieval over curated statutory knowledge base.
2. **Relevance & Confidence Gating (RCG)**: Dual-check verification gating out spurious legal noise based on essential ingredients (*Actus Reus*, *Mens Rea*, weapon brandishing).
3. **Prefix Attention Fusion (PFL)**: Guided attention prefix over `law-ai/InLegalBERT` transformer backbone.
4. **Multi-Head Prediction Network (MHPN)**:
   - **Head 1**: Law Articles (Bidirectional IPC 1860 & BNS 2023)
   - **Head 2**: Criminal Charge Categories (12 Indian Offence Classes)
   - **Head 3**: Statutory Penalty Severity Tiers (Tiers 1 to 4)
5. **Judicial Consistency Verifier (JCV)**: Enforces statutory ceiling/floor constraints, Section 300 Exception 4 conversion rules, and computes the Judicial Consistency Index (JCI).
